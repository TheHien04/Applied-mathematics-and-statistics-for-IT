## **Giải thuật thực hiện**

### 1. **Khử Gauss (Forward Elimination)**
- Đưa ma trận mở rộng \( A \) về dạng bậc thang bằng cách:
  - Tìm phần tử **pivot** khác 0 ở từng cột.
  - Đổi hàng để đưa phần tử pivot lên trên.
  - Chuẩn hóa hàng sao cho phần tử đầu tiên của hàng = 1.
  - Dùng phép biến đổi sơ cấp để khử tất cả các phần tử phía dưới pivot.

### 2. **Thế ngược (Back Substitution)**
- Bắt đầu từ hàng cuối (ẩn cuối cùng), thế ngược lên trên để tìm nghiệm từng biến.
- Kiểm tra số nghiệm:
  - Nếu có hàng toàn 0 nhưng vế phải khác 0 → **vô nghiệm**
  - Nếu số phương trình độc lập < số biến → **vô số nghiệm**
  - Nếu số phương trình độc lập = số biến → **nghiệm duy nhất**

---



```python
def Gauss_elimination(A):
    n = len(A)
    m = len(A[0])
    
    for i in range(n):
        # Tìm hàng có phần tử khác 0 đầu tiên tại cột i (pivoting đơn giản)
        max_row = i
        for k in range(i+1, n):
            if abs(A[k][i]) > abs(A[max_row][i]):
                max_row = k
        A[i], A[max_row] = A[max_row], A[i]

        # Nếu phần tử pivot = 0 thì bỏ qua (có thể là vô số nghiệm hoặc vô nghiệm)
        if A[i][i] == 0:
            continue

        # Chuẩn hóa hàng i
        pivot = A[i][i]
        A[i] = [x / pivot for x in A[i]]

        # Khử các hàng phía dưới
        for j in range(i+1, n):
            ratio = A[j][i]
            A[j] = [A[j][k] - ratio * A[i][k] for k in range(m)]

    return A
```

```python
def back_substitution(A):
    n = len(A)
    m = len(A[0])
    x = [0] * (m - 1)

    # Kiểm tra vô nghiệm
    for i in range(n):
        if all(abs(A[i][j]) < 1e-10 for j in range(m-1)) and abs(A[i][-1]) > 1e-10:
            return "Hệ phương trình vô nghiệm."

    # Đếm số phương trình độc lập
    rank = 0
    for i in range(n):
        if any(abs(x) > 1e-10 for x in A[i][:-1]):
            rank += 1

    if rank < m - 1:
        return "Hệ phương trình có vô số nghiệm."

    # Giải bằng thế ngược
    for i in reversed(range(n)):
        x[i] = A[i][-1]
        for j in range(i+1, m-1):
            x[i] -= A[i][j] * x[j]
    return x
```

```python
# Ví dụ hệ phương trình: 
# x + y + z = 6
# 2y + 5z = -4
# 2x + 5y - z = 27

A = [
    [1, 1, 1, 6],
    [0, 2, 5, -4],
    [2, 5, -1, 27]
]

A_step = Gauss_elimination([row[:] for row in A])  # dùng copy để không làm thay đổi ma trận gốc
print("Ma trận sau Gauss:")
for row in A_step:
    print(row)

result = back_substitution([row[:] for row in A_step])
print("\nNghiệm hệ phương trình:")
print(result)
```

```python
import numpy as np

A_np = np.array([row[:-1] for row in A])
b_np = np.array([row[-1] for row in A])

try:
    sol = np.linalg.solve(A_np, b_np)
    print("\nNghiệm dùng numpy.linalg.solve:")
    print(sol)
except np.linalg.LinAlgError:
    print("\nHệ phương trình vô nghiệm hoặc vô số nghiệm.")
```

## **Ý tưởng thực hiện**

Hệ phương trình tuyến tính được giải bằng phương pháp khử Gauss bao gồm hai giai đoạn:
1. **Khử tiến (forward elimination):** đưa hệ phương trình về dạng bậc thang.
2. **Thế ngược (back substitution):** tìm nghiệm từ hệ bậc thang đó.

Trong bài này, sinh viên thực hiện toàn bộ quá trình mà không sử dụng các hàm giải hệ phương trình có sẵn trong thư viện Python như `numpy.linalg.solve`.

---

## **Mô tả các hàm**

### `Gauss_elimination(A)`
- **Input:** Ma trận mở rộng của hệ \( Ax = b \) (dạng list of lists).
- **Output:** Ma trận có dạng bậc thang hàng.
- **Xử lý:** Lần lượt chọn pivot cho mỗi cột, đổi hàng, chia chuẩn, rồi khử các hàng bên dưới bằng phép biến đổi sơ cấp.

### `back_substitution(A)`
- **Input:** Ma trận bậc thang hàng.
- **Output:** Trả về:
  - Danh sách nghiệm nếu có nghiệm duy nhất.
  - Thông báo nếu vô nghiệm hoặc vô số nghiệm.
- **Xử lý:** Thực hiện thế ngược từ hàng cuối lên, kiểm tra trường hợp đặc biệt như hàng 0, hoặc số phương trình độc lập.
